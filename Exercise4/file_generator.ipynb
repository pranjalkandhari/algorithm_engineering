```python
import os
```

```python
files = os.listdir('Data')
files
for file in files:
    print('\"' + file + '\",')
```

```python
for file in files:
    if (file[0] == '0' or file[0] == '1' or file[0] == '2'):
        print(file)        
        line_number = 0
        data_line = False
        count_points = 0
        n = 0
        with open('Data/' + file, 'r') as f:
            write_file_name = file.split('.')[0] + '.txt'
            with open('input_data/' + write_file_name, 'w') as f_write:
                # Iterate through each line in the file
                for line in f:
                    # Print the current line
                    new_line = (line.strip())  # strip() removes the newline character at the end of each line
                    # print(line)

                    # Code to generate text file.
                    li_split  = line.split()
                    if ('DIMENSION' in li_split[0]):
                        print(line)
                        n = int(li_split[len(li_split)-1])
                        print('n:', n)
                        f_write.write(str(n) + '\n')


                    if (li_split[0] == 'EOF'):
                        continue

                    if (data_line == True):
                        if (len(li_split) != 3):
                            print("ALERT!")
                            print(line)
                        else:
                            count_points += 1
                            # Here we get a list with 3 numbers. We can put it in 3 numbers.
                            x = float(li_split[1])
                            y = float(li_split[2])
                            f_write.write(str(x) + ' ' + str(y) + '\n')

                    if ('NODE_COORD_SECTION' in li_split[0]):
                        data_line = True

                    line_number += 1
                print("n, count_points:",n, count_points)
                print('---------------------------')
```
